```python
import pyspark.sql.functions as F
```

```python
# ISO date format - 2024-10-27T13:00:00 (yyyy-MM-ddTHH:mm:ss)
# Read Date as Input and get day of the week - Sun, Monday...etc
# Create input widget
dbutils.widgets.text("input_date","")
```

```python
# Read value from the widget
input_date = dbutils.widgets.get("input_date")
print(input_date)
```

```python
# Formatting the input date to extract day of the week
# Set conf to LEGACY to support datetime patterns
spark.conf.set("spark.sql.legacy.timeParserPolicy","LEGACY")
input_day = spark.sql(
    f"""
        SELECT date_format(to_timestamp('{input_date}',"yyyy-MM-dd'T'HH:mm:ss"),'E' )
    
    """
).collect()[0][0]

print(input_day)
```

```python
# Export the day of the week using task Values to use in next task
dbutils.jobs.taskValues.set("input_day",input_day)
```
